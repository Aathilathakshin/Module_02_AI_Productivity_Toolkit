## 1. Prompt Engineering

### Zero-shot Prompting
Giving AI a direct instruction without providing examples.

### Few-shot Prompting
Providing examples before asking AI to perform a task.

```python
# Zero-shot prompting example

prompt = "Explain Generative AI in one sentence."
print(prompt)
```

## 2. Structured Output

Structured output means asking AI to return information
in a specific format, such as JSON.

```python
import json

data = [
    {
        "feature": "Syntax",
        "python": "Simple and readable",
        "java": "Structured and more detailed"
    },
    {
        "feature": "Execution",
        "python": "Uses an interpreter",
        "java": "Uses the JVM to execute bytecode"
    },
    {
        "feature": "Output",
        "python": "Uses print()",
        "java": "Uses System.out.println()"
    }
]

print(json.dumps(data, indent=2))
```

## 3. Summarization

Summarization converts a long paragraph into a shorter
version while keeping the important information.

## 4. Classification

Classification assigns text or data to predefined categories.

```python
# Simple classification example

message = "I love learning Python"

if "love" in message.lower():
    category = "Positive"
else:
    category = "Other"

print("Category:", category)
```

## 5. AI-assisted Coding and Debugging

Debugging is the process of identifying and fixing errors
in a program.

```python
# Buggy code

num1 = 10
num2 = 20
num3 = 30

average = num1 + num2 + num3 / 3

print("Average:", average)
```

```python
# Corrected code

num1 = 10
num2 = 20
num3 = 30

average = (num1 + num2 + num3) / 3

print("Average:", average)
```
